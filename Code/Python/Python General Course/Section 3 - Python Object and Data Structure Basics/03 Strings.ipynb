### String Basics and Printing
This section demonstrates basic string creation and printing in Python.

```python
string1 = "Hello"
print(string1[0] + " is the letter H")
```

### String Indexing
This cell demonstrates how to access individual characters in a string using indexing.

```python
last_char = string1[-1]
print(last_char)
```

### String Slicing
This cell shows how to extract substrings from a string using slicing.

```python
my_slice = string1[1:4]
print(my_slice)
```

### Advanced Slicing
This cell demonstrates more advanced string slicing, such as skipping characters using a step value.

```python
longer_string = "This is a longer string example"
every_other = longer_string[1:-1:2]
print(every_other)
```

### String Concatenation and Immutability
This cell explains how to concatenate strings and discusses the immutability of Python strings.

```python
a = "hello"
b = "I'm"
c = "Jude"
d = a + " " + b + " " + c
print(d)
```

### Practice or Additional Example
This cell is for further practice or an additional example related to string operations.

```python
escape = "hello \nworld"
print(escape)
escape = "hello \tworld"
print(escape)
escape = "She said: \"Hello!\""
print(escape)
escape = "Backslash: \\ and single quote: \\'"
print(escape)
escape = "Carriage return: hello\rworld"
print(escape)
escape = "Bell sound (may not work everywhere): hello\aworld"
print(escape)
```

### String Length Check
Counts the number of characters in the current string using `len()`.

```python
length = len(escape)
print(length)
```

### Useful Methods

```python
x = "Hello World"
print("capitalize():", x.capitalize(), "- First character capitalized.")
print("casefold():", x.casefold(), "- Aggressive lowercase for caseless matching.")
print("center(20):", x.center(20), "- Centered in width 20.")
print("count('l'):", x.count('l'), "- Count of 'l'.")
print("encode():", x.encode(), "- Encoded to bytes.")
print("endswith('World'):", x.endswith('World'), "- Ends with 'World'.")
print("expandtabs(4):", 'Hello\tWorld'.expandtabs(4), "- Tab expanded to spaces.")
print("find('o'):", x.find('o'), "- Lowest index of 'o'.")
print("format():", 'Hello {0}'.format('World'), "- Format with placeholders.")
print("format_map():", 'Hello {name}'.format_map({'name':'World'}), "- Format with mapping.")
print("index('W'):", x.index('W'), "- Index of 'W'.")
print("isalnum():", x.isalnum(), "- All alphanumeric?")
print("isalpha():", x.isalpha(), "- All alphabetic?")
print("isascii():", x.isascii(), "- All ASCII?")
print("isdecimal():", '123'.isdecimal(), "- All decimal?")
print("isdigit():", '123'.isdigit(), "- All digits?")
print("isidentifier():", 'hello_world'.isidentifier(), "- Valid identifier?")
print("islower():", x.islower(), "- All lowercase?")
print("isnumeric():", '123'.isnumeric(), "- All numeric?")
print("isprintable():", x.isprintable(), "- All printable?")
print("isspace():", '   '.isspace(), "- All whitespace?")
print("istitle():", x.istitle(), "- Title case?")
print("isupper():", x.isupper(), "- All uppercase?")
print("join(['A','B']):", x.join(['A','B']), "- Join with string as separator.")
print("ljust(15):", x.ljust(15, '-'), "- Left-justified, padded with '-'.")
print("lower():", x.lower(), "- Lowercase.")
print("lstrip('H'):", x.lstrip('H'), "- Leading 'H' removed.")
print("maketrans() & translate():", x.translate(str.maketrans('H', 'J')), "- 'H' replaced with 'J'.")
print("partition(' '):", x.partition(' '), "- Split at first space.")
print("removeprefix('Hello '):", x.removeprefix('Hello '), "- Remove prefix if present.")
print("removesuffix(' World'):", x.removesuffix(' World'), "- Remove suffix if present.")
print("replace('World','Python'):", x.replace('World','Python'), "- Replace substring.")
print("rfind('l'):", x.rfind('l'), "- Highest index of 'l'.")
print("rindex('l'):", x.rindex('l'), "- Highest index of 'l'.")
print("rjust(15):", x.rjust(15, '-'), "- Right-justified, padded with '-'.")
print("rpartition(' '):", x.rpartition(' '), "- Split at last space.")
print("rsplit('l'):", x.rsplit('l'), "- Split from right at 'l'.")
print("rstrip('d'):", x.rstrip('d'), "- Trailing 'd' removed.")
print("split():", x.split(), "- Split into list.")
print("splitlines():", 'Hello\nWorld'.splitlines(), "- Split at line breaks.")
print("startswith('Hello'):", x.startswith('Hello'), "- Starts with 'Hello'.")
print("strip('Hd'):", x.strip('Hd'), "- Leading/trailing 'H'/'d' removed.")
print("swapcase():", x.swapcase(), "- Swap case.")
print("title():", x.title(), "- Title case.")
print("upper():", x.upper(), "- Uppercase.")
print("zfill(15):", x.zfill(15), "- Pad left with zeros to width 15.")
```

### String formatting  
#### .format() method

```python
print("This is a string {}".format('Inserted'))
```

### Basic `.format()` Insertion
Inserts one value into a placeholder using `.format()`.

```python
print('The {2} {1} {0}'.format('fox','brown','quick'))
```

### Positional `.format()` Ordering
Shows how numbered placeholders can reorder inserted values.

```python
print('The {q} {b} {f}'.format(f='fox',b='brown',q='quick'))
```

### Keyword `.format()` Arguments
Uses named placeholders for clearer formatting.

```python
print("The result was {r:5.3f}".format(r=result))
```
