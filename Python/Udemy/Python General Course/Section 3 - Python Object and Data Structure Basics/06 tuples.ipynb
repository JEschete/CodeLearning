# Tuples in Python
This notebook summarizes tuple concepts from Section 3 notes with simple examples.

## Learning goals
- Understand tuple syntax
- Compare tuples vs lists
- Use tuple methods: `count()` and `index()`
- See why immutability matters

## 1) Tuple syntax and type checks
Tuples use parentheses `()` and are similar to lists, but immutable.

```python
t = (1, 2, 3)
mylist = [1, 2, 3]

print('t      =', t)
print('mylist =', mylist)
print('type(t):', type(t))
print('type(mylist):', type(mylist))
```

## 2) Tuple basics
Like lists, tuples support length checks, mixed data types, indexing, negative indexing, and slicing.

```python
mixed_t = ('one', 2, 3.5, True)

print('mixed_t:', mixed_t)
print('len(mixed_t):', len(mixed_t))
print('mixed_t[0]:', mixed_t[0])
print('mixed_t[-1]:', mixed_t[-1])
print('mixed_t[1:3]:', mixed_t[1:3])
```

## 3) Tuple methods: `count()` and `index()`
Tuples have fewer methods than lists. Two key methods are:
- `count(value)` = number of occurrences
- `index(value)` = first position of the value

```python
letters_t = ('a', 'a', 'b')

print('letters_t:', letters_t)
print("letters_t.count('a'):", letters_t.count('a'))
print("letters_t.index('a'):", letters_t.index('a'))
print("letters_t.index('b'):", letters_t.index('b'))
```

## 4) Immutability (main difference from lists)
Lists can be reassigned by index; tuples cannot.

```python
mutable_list = [1, 2, 3]
immutable_tuple = (1, 2, 3)

mutable_list[0] = 'NEW'
print('List after reassignment:', mutable_list)

try:
    immutable_tuple[0] = 'NEW'
except TypeError as e:
    print('Tuple reassignment error:', e)
```

## 5) Nested access pattern
Tuples can be part of nested structures, and you can chain indexing/method calls.

```python
nested = {'key1': ['a', 'b', 'c']}

uppercase_c = nested['key1'][2].upper()
print('nested:', nested)
print('nested[\'key1\'][2].upper():', uppercase_c)
```
