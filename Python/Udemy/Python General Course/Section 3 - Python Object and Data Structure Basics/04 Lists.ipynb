```python
my_list3 = ['one', 'two', 'three']
```

### Index Into a List
Accesses the first element using index `0`.

```python
my_list[0]
```

### Slice From an Index
Returns all elements from index `1` to the end.

```python
my_list[1:]
```

### Slice Up to the Last Item
Returns all elements except the final one using `[:-1]`.

```python
my_list[:-1]
```

```python
my_list + my_list2 + my_list3
```

```python
new_list = my_list + my_list2 + my_list3
```
