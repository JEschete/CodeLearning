# Sets in Python
This notebook summarizes set concepts from Section 3 notes with practical examples.

## Learning goals
- Understand what sets are
- Add elements with `add()`
- See why duplicates are removed
- Convert lists to sets for unique values

## 1) Create an empty set
Use `set()` to create an empty set.

```python
myset = set()
print('Empty set:', myset)
print('Type:', type(myset))
```

## 2) Add elements with `add()`
Sets keep only unique values, so duplicate inserts are ignored.

```python
myset.add(1)
myset.add(2)
myset.add(2)  # duplicate, should be ignored

print('Set after adds:', myset)
print('Length (unique count):', len(myset))
```

## 3) Sets are unordered
Sets do not maintain index-based ordering like lists.

```python
print('Current set display:', myset)
print('Contains 1?', 1 in myset)
print('Contains 99?', 99 in myset)
```

## 4) Convert list to set (remove duplicates)
A common beginner use case for sets is deduplicating data.

```python
mylist = [1, 1, 1, 2, 2, 2, 3, 3, 3]
unique_values = set(mylist)

print('Original list:', mylist)
print('Unique set:', unique_values)
print('Unique values as sorted list:', sorted(unique_values))
```

## 5) Quick check
Try these yourself:
1. Create a set from `['a', 'a', 'b', 'c', 'c']`.
2. Add `'d'` and `'d'` again.
3. Check membership for `'b'` and `'z'`.

```python
# Optional practice starter
letters = ['a', 'a', 'b', 'c', 'c']
letters_set = set(letters)

letters_set.add('d')
letters_set.add('d')

print('letters_set:', letters_set)
print("'b' in set:", 'b' in letters_set)
print("'z' in set:", 'z' in letters_set)
```

## 6) Key takeaway
Sets are best when you need **unique values** and do **not** need index-based ordering.
